## Significado de los símbolos

- **✓ SOLUCIONADO:** existe una medida implementada y una evidencia verificable.
- **✗ PENDIENTE:** falta implementación, trámite o comprobación. Impide declarar el sistema listo cuando afecta producción.

> Un diseño o una decisión escrita no reemplaza la implementación. Por eso Temurin instalado y Temurin configurado en Spring Boot/VPS aparecen como controles separados.

## Criterio para cambiar una ✗ por ✓

Un elemento solo se marca como solucionado cuando cumple las cuatro condiciones:

1. La medida está implementada, no solamente propuesta.
2. Existe una prueba o documento que demuestra que funciona.
3. La evidencia está vinculada en la columna correspondiente o en la bitácora.
4. No depende de una tarea pendiente que pueda anular la protección.

Ejemplos de evidencia válida: commit, prueba automatizada, captura de configuración, constancia RNPD, contrato/DPA, acta firmada, reporte de auditoría o restauración comprobada.
