## Cómo implementar el Libro de Reclamaciones

No hay que ir a la municipalidad, notaría, SUNAT ni a una oficina de Indecopi para “registrarlo”. El Libro no requiere legalización, certificación notarial ni registro previo ante Indecopi.

1. Confirmar razón social/nombre legal, RUC, dirección y responsable.
2. Entrar a [Tu Libro de Reclamaciones de Indecopi](https://consumidor.indecopi.gob.pe/tulibro/#/).
3. Generar el formato oficial físico o virtual y descargar el aviso.
4. Crear un libro para cada establecimiento con código único, por ejemplo `LOCAL-01` y `LOCAL-02`.
5. Colocar el aviso visible y mantener el libro disponible cuando un consumidor lo solicite.
6. Entregar al consumidor su copia de la hoja o constancia electrónica.
7. Designar a la persona que responderá por escrito, normalmente por correo, en un máximo de 15 días hábiles improrrogables.
8. Conservar hoja, respuesta y constancia de entrega por al menos dos años.
9. En la página, colocar un acceso visible y permanente al Libro virtual y probarlo en móvil/escritorio.
10. Consultar con contador/legal si corresponde reportar al SIREC; la FAQ oficial lo exige a proveedores con ingresos anuales iguales o mayores a 3000 UIT.

Indecopi puede actualizar el reglamento. Debe revisarse nuevamente la fuente oficial justo antes de producción.

## Venta de productos por la web, explicado correctamente

| Situación | Consecuencia |
|---|---|
| Mostrar un producto en el catálogo | No genera por sí solo una venta ni un impuesto nuevo |
| El cliente paga por Izipay y recoge en el local | Sí existe una venta; debe emitirse el comprobante y declararse según el régimen del negocio |
| El proveedor entregó boleta/factura al comprar el producto | Ayuda a acreditar procedencia y costo, pero no reemplaza la NSO ni la boleta que Rubí entrega a su cliente |
| El producto es cosmético | Antes de publicarlo se verifica en el rotulado su NSO, lote, origen, responsable y vencimiento |
| Rubí solo revende un producto formal | Normalmente verifica la NSO del producto/proveedor; no se presume que deba tramitar una NSO nueva a su nombre |
| Falta NSO o rotulado confiable | El sistema no permite publicarlo hasta que Rubí/proveedor acredite la información |
| Se devuelve o anula la venta | Izipay y el comprobante tributario requieren flujos relacionados, pero diferentes |

La aplicación guardará producto, lote y vencimiento cuando sea útil para trazabilidad, pero Rubí debe definir con su contador y proveedor qué documentos conserva.
