# Responsabilidades para llevar el sistema a producción

**Proyecto:** Rubí Estética Integral  
**Última revisión:** 30 de agosto de 2026  
**Objetivo:** separar claramente lo que corresponde al desarrollador, a Rubí y a los asesores externos.

La empresa indicó que sus locales están equipados, cuentan con licencias y trabajan con especialistas certificadas. La tarea pendiente no es asumir que eso está mal, sino reunir la evidencia y relacionar cada tratamiento con el personal legalmente autorizado. Los practicantes y personal part-time solo deben recibir en el sistema las funciones que Rubí y su asesor confirmen.

## SUNAT explicado sin mezclarlo con reclamos

| Pregunta | Respuesta para este proyecto | Acción |
|---|---|---|
| ¿Se paga un impuesto nuevo solo por tener página web? | **No por el simple hecho de tener la web.** La web es otro canal de venta. Las ventas y servicios que genere forman parte de los ingresos del negocio y se declaran según su RUC y régimen tributario. | El contador confirma régimen, declaraciones y pagos actuales. |
| ¿SUNAT administra los reclamos? | No. El Libro de Reclamaciones corresponde a protección del consumidor e Indecopi. | Separar el flujo de reclamos del flujo tributario. |
| ¿Qué exige SUNAT al vender por Internet? | La transacción por Internet no elimina la obligación de emitir comprobante y declarar/pagar lo que corresponda. | Emitir boleta al consumidor final o factura cuando corresponda. |
| ¿Izipay reemplaza la boleta? | No. Izipay procesa/cobra el pago; el comprobante tributario es otro documento. | Vincular pago, pedido y número de comprobante. |
| ¿La web debe conectarse directamente con SUNAT? | No necesariamente al inicio. Puede conservarse el sistema actual de emisión, utilizar SOL/App SUNAT, un PSE/OSE o integrar el sistema propio más adelante. | Elegir con el contador antes de programar. |
| ¿Podemos guardar la Clave SOL? | Nunca en Angular, Git ni una tabla común. Una integración formal usa credenciales y certificados protegidos según el sistema elegido. | Diseñar el backend después de elegir modalidad. |
| ¿Qué ocurre con los dos locales? | SUNAT indica que quienes desarrollan actividades en más de un establecimiento deben declararlos en el RUC como establecimientos anexos. La distancia entre ellos no elimina ese deber. | El contador verifica ambas direcciones en la ficha RUC. |

**Conclusión:** antes de programar facturación se pregunta a Rubí: “¿Con qué sistema emite hoy sus boletas y facturas y bajo qué régimen está?”. No se inventa un flujo tributario desde el código.

## Evaluación del plan Hostinger KVM 2 mostrado

Datos leídos de la captura: **2 vCPU, 8 GB RAM, 100 GB NVMe y 8 TB de transferencia**; precio promocional mostrado de **US$ 9,19/mes** y renovación de **US$ 17,99/mes** durante la condición indicada. El precio y periodo deben verificarse nuevamente en el checkout antes de contratar.

| Aspecto | Evaluación | Responsable real |
|---|---|---|
| Capacidad inicial | Adecuada como punto de partida para Nginx, Angular, Spring Boot y MySQL con tráfico inicial bajo/moderado. Es una estimación; se confirma midiendo RAM, CPU, disco y consultas. | Desarrollo monitorea; Rubí aprueba escalado |
| Infraestructura física | Hostinger administra hardware, red y disponibilidad del centro de datos. | Hostinger |
| Firewall externo | hPanel permite filtrar tráfico antes de llegar al VPS. Deben abrirse únicamente 80/443 y SSH restringido. | Desarrollo configura |
| Firewall del sistema | El firewall de Hostinger no reemplaza todas las reglas del sistema operativo ni garantiza detener un DDoS volumétrico. | Desarrollo configura UFW/nftables |
| Malware y panel | Hostinger ofrece escáner de malware y herramientas de seguridad/monitoreo. | Hostinger ofrece; desarrollo revisa alertas |
| Administración | Hostinger confirma que sus VPS son **autogestionados**. Parches, usuarios, Nginx, Java, MySQL y seguridad de la aplicación no vienen resueltos automáticamente. | Desarrollo/DevOps |
| Respaldo | El VPS tiene respaldo automático semanal por defecto. No alcanza para citas y pagos diarios. | Desarrollo crea copia diaria cifrada externa |
| Protección absoluta | Ningún proveedor puede garantizar que un sistema sea imposible de hackear. Se reduce el riesgo con capas, actualización, mínimo privilegio, monitoreo y respuesta. | Compartida |
| Datos fuera de Perú | Debe elegirse el centro de datos y documentar DPA/transferencia internacional. | Rubí/legal + desarrollo |

### Decisión recomendada

**Hostinger KVM 2 es aceptable para el inicio** si se contrata a nombre de Rubí y aplicamos todo el endurecimiento, respaldo externo y monitoreo. Es económico, pero exige mantenimiento técnico continuo.

Si Rubí prefiere pagar más para reducir la administración de MySQL, una alternativa es una instancia y base administrada en Amazon Lightsail. AWS gestiona parches menores, cifrado, respaldo y restauración a un punto en el tiempo de la base; el plan administrado empieza oficialmente en US$ 15/mes, aparte del servidor. Antes de elegirlo hay que comprobar la versión de MySQL y la región disponibles. Aun así, la aplicación, permisos, privacidad y respaldo independiente siguen siendo responsabilidad nuestra.

**No comprar todavía** hasta confirmar: titular de la cuenta, renovación real, presupuesto anual, país del centro de datos, DPA y quién realizará el mantenimiento mensual.

## Orden de trabajo

1. **Reunión con Rubí:** completar C01–C18 con documentos, responsables y respuestas reales.
2. **Contador:** confirmar establecimientos anexos, régimen, comprobantes y sistema de emisión actual.
3. **Privacidad/legal:** RNPD, política, consentimiento, conservación, DPA y Libro de Reclamaciones.
4. **Sanitario:** documentar especialistas, practicantes y tratamientos autorizados.
5. **Backend:** Temurin 25, Spring Security, roles, auditoría, MySQL y pagos de prueba.
6. **Preproducción:** pruebas funcionales, seguridad, restauración e incidente simulado.
7. **Hosting:** contratar solamente después de cerrar titularidad, ubicación y presupuesto.
8. **Producción:** Rubí y responsables firman la aceptación; recién entonces se cargan datos y pagos reales.
