# Plan de preparacion y despliegue en Hostinger VPS

**Proyecto:** Rubi Estetica Integral  
**Fecha:** 30/08/2026  
**Objetivo:** dejar una guia clara para preparar, comprar, configurar y publicar el sistema con Angular, Spring Boot y MySQL/MariaDB sin perder datos y reduciendo riesgos legales, tecnicos y de seguridad.

> Nota importante: en el VPS no se instala Eclipse. Eclipse, IntelliJ o Visual Studio Code son herramientas para desarrollar en la computadora local. En produccion se instala Java, Nginx, MySQL/MariaDB, firewall, certificados SSL y servicios del sistema.

## 1. Arquitectura objetivo

| Parte | Donde queda | Funcion |
|---|---|---|
| Angular | VPS, servido por Nginx | Pagina visual para clientes, reservas, productos y paneles |
| Spring Boot | VPS, como servicio Java | API del sistema: login, reservas, pacientes, productos, pagos, reclamos |
| MySQL/MariaDB | VPS o base administrada | Datos reales: usuarios, pacientes, citas, tratamientos, ventas, reclamos |
| Dominio .com | DNS apuntando al VPS | Entrada publica del negocio |
| HTTPS/SSL | Nginx + certificado | Conexion segura con candado |
| Backups | VPS y/o almacenamiento externo | Recuperacion ante errores, perdida o dano de datos |

## 2. Antes de comprar Hostinger o el dominio

### 2.1 Revisar licencias y dependencias

| Punto | Estado esperado | Responsable |
|---|---|---|
| Java/JDK | Usar Eclipse Temurin LTS, no Oracle JDK comercial | Desarrollador |
| Angular | Revisar `npm audit` y dependencias sin vulnerabilidades criticas | Desarrollador |
| Librerias frontend | Confirmar que sean compatibles con uso comercial | Desarrollador |
| Imagenes/videos | Confirmar que pertenecen a Rubi o que tienen autorizacion | Cliente |
| Productos cosmeticos | Confirmar que son productos permitidos y con documentacion del proveedor | Cliente |
| Izipay online | Confirmar que tiene o puede contratar integracion ecommerce/API | Cliente + desarrollador |

### 2.2 Confirmar datos legales del negocio

- RUC.
- Razon social.
- Nombre comercial.
- Direccion exacta de los dos locales.
- Confirmar si ambos locales trabajan bajo el mismo RUC.
- Correo oficial.
- WhatsApp o telefono oficial.
- Reglas de reserva, adelanto, reprogramacion y no asistencia.
- Confirmar si existe libro de reclamaciones fisico.
- Confirmar si se enviaran promociones por correo, SMS o WhatsApp.

## 3. Documentos legales que deben prepararse antes de publicar

| Documento | Que debe contener | Quien aporta informacion | Quien implementa |
|---|---|---|---|
| Politica de privacidad | Datos que se recopilan, finalidad, acceso, seguridad, derechos ARCO, contacto | Cliente + desarrollador | Desarrollador |
| Terminos y condiciones | Reservas, pagos, puntualidad, reprogramaciones, promociones, productos, recojo en local | Cliente | Desarrollador |
| Politica de cookies | Cookies necesarias, analiticas, publicidad, aceptar/rechazar/configurar | Desarrollador | Desarrollador |
| Libro de reclamaciones virtual | Formulario, codigo, copia al cliente, registro interno | Cliente | Desarrollador |
| Consentimiento promocional | Autorizacion separada para correo, SMS o WhatsApp | Cliente + desarrollador | Desarrollador |
| Consentimiento de imagen | Fotos o videos de pacientes si se publican | Cliente | Cliente + desarrollador si va en web |

**Derechos ARCO:** Acceso, Rectificacion, Cancelacion y Oposicion. En simple: el cliente puede pedir ver, corregir, cancelar o limitar el uso de sus datos cuando corresponda.

## 4. Banco de datos personales

Como el sistema guardara datos de pacientes/clientes, Rubi o su representante debe inscribir el banco de datos personales ante la Autoridad Nacional de Proteccion de Datos Personales.

No se entrega la base MySQL completa. Normalmente se declara la existencia del banco de datos y se describe:

| Campo | Ejemplo para el proyecto |
|---|---|
| Nombre del banco | Clientes y pacientes de Rubi Estetica Integral |
| Titular | Razon social o titular del negocio |
| Finalidad | Reservas, tratamientos, historial, ventas, reclamos y atencion al cliente |
| Datos tratados | DNI, nombres, apellidos, celular, correo, citas, tratamientos, pagos, reclamos |
| Personas registradas | Clientes, pacientes y usuarios web |
| Ubicacion | Servidor VPS/base de datos en nube |
| Seguridad | Roles, HTTPS, hash de contrasenas, base privada, backups, firewall |
| Proveedores | Hosting, pasarela de pago, correo/SMS si se usan |

El desarrollador prepara la informacion tecnica. La titular del negocio realiza o autoriza el tramite.

## 5. Seguridad que debe implementarse en Spring Boot y MySQL

### 5.1 Roles

| Rol | Acceso permitido |
|---|---|
| Administradora | Gestion completa: ingresos, pagos, pacientes, reportes, configuracion |
| Recepcion | Registrar citas, ver lista del dia, registrar llegada/estado, registrar pedidos |
| Especialista | Ver atenciones asignadas y seguimiento necesario |
| Cliente | Ver su cuenta, reservas, historial permitido y compras |

### 5.2 Contrasenas

- No guardar contrasenas en texto plano.
- No guardar contrasenas en Angular.
- Usar hash seguro como BCrypt o Argon2.
- Si el usuario olvida su clave, se restablece; no se recupera la clave anterior.

### 5.3 Base de datos privada

- MySQL/MariaDB no debe estar abierto al publico por el puerto 3306.
- Spring Boot debe ser el unico que se conecta a la base.
- Crear usuario de base de datos limitado, no usar root desde la aplicacion.
- Guardar credenciales en variables de entorno, no en GitHub.
- Firewall activo.
- Backups diarios.
- Prueba de restauracion de backups.

## 6. Compra y preparacion del VPS

### 6.1 Comprar

- Comprar dominio `.com`.
- Comprar VPS en Hostinger o proveedor equivalente.
- Elegir sistema operativo LTS, por ejemplo Ubuntu LTS.
- Guardar credenciales iniciales de forma segura.
- Activar backups del proveedor si el plan lo permite.

### 6.2 Configuracion inicial del VPS

1. Crear usuario administrador del servidor.
2. Configurar acceso SSH seguro.
3. Actualizar paquetes del sistema.
4. Instalar firewall.
5. Abrir solo puertos necesarios: 80, 443 y SSH.
6. Instalar Java Temurin LTS.
7. Instalar MySQL o MariaDB.
8. Instalar Nginx.
9. Instalar Certbot para SSL/HTTPS.
10. Crear carpetas para frontend, backend, logs y backups.

## 7. Despliegue del backend Spring Boot

1. Construir el backend localmente:

```bash
mvn clean package
```

2. Subir el archivo `.jar` al VPS.
3. Crear variables de entorno para:

- URL de base de datos.
- Usuario de base de datos.
- Contrasena de base de datos.
- Secret/JWT.
- Credenciales de Izipay si se integra pago online.
- Configuracion de correo/SMS si se usa.

4. Crear servicio `systemd` para que Spring Boot arranque automaticamente.
5. Revisar logs.
6. Probar endpoints internos antes de abrir la web al publico.

## 8. Despliegue del frontend Angular

1. Revisar que la URL de API apunte al dominio real o endpoint de produccion.
2. Construir Angular:

```bash
npm run build
```

3. Subir la carpeta generada en `dist/` al VPS.
4. Configurar Nginx para servir Angular.
5. Configurar Nginx para redirigir `/api` hacia Spring Boot.
6. Activar HTTPS.
7. Probar navegacion, recarga de rutas, login, reservas y paneles.

## 9. Migracion de base de datos sin perder informacion

Si ya existe una base local con datos reales o de configuracion, se exporta e importa.

### 9.1 Exportar local

```bash
mysqldump -u usuario -p rubi_db > rubi_db_backup.sql
```

### 9.2 Importar en VPS

```bash
mysql -u usuario -p rubi_db < rubi_db_backup.sql
```

### 9.3 Validar despues de importar

- Contar usuarios.
- Contar pacientes.
- Contar productos.
- Contar tratamientos.
- Probar reservas.
- Probar login.
- Probar pagos o estados de pago.
- Probar reclamos.

> Si actualmente los datos estan quemados en Angular, arrays `.ts`, `localStorage` o archivos estaticos, esos datos deben migrarse a MySQL o registrarse nuevamente desde el panel.

## 10. Backups

Como el sistema manejara reservas y ventas, el backup debe ser diario.

| Backup | Frecuencia sugerida | Retencion sugerida |
|---|---|---|
| Base de datos MySQL/MariaDB | Diario | 7 dias diarios + 4 semanales |
| Archivos subidos | Diario o segun cambios | 7 dias diarios + 4 semanales |
| Configuracion del servidor | Cada cambio importante | Mantener version documentada |

Tambien debe hacerse una prueba de restauracion. Un backup que nunca se prueba puede fallar el dia que realmente se necesita.

## 12. Riesgos principales y como reducirlos

| Riesgo | Impacto | Prevencion |
|---|---|---|
| Usar Oracle JDK comercial sin revisar licencia | Legal/costos | Usar Eclipse Temurin LTS |
| Datos personales sin politica clara | Legal/Indecopi/datos personales | Politica de privacidad y banco de datos |
| Contrasenas en texto plano | Seguridad critica | Hash BCrypt/Argon2 |
| MySQL abierto a internet | Seguridad critica | Firewall y acceso solo desde backend |
| Credenciales en GitHub | Seguridad critica | Variables de entorno y `.gitignore` |
| Sin backups | Perdida de datos | Backups diarios y prueba de restauracion |
| Cookies de publicidad sin consentimiento | Legal/privacidad | Banner aceptar/rechazar/configurar |
| Promociones sin autorizacion | Legal/consumidor/datos | Check separado y opcional |
| Libro de reclamaciones ausente | Riesgo Indecopi | Libro fisico/virtual visible |
| Publicar sin pruebas | Fallas operativas | Checklist de QA antes de produccion |

## 13. Orden recomendado de trabajo

1. Terminar frontend limpio y responsive.
2. Confirmar datos legales y reglas con Rubi.
3. Definir estructura real de base de datos.
4. Crear backend Spring Boot.
5. Implementar seguridad, roles y hash de contrasenas.
6. Migrar datos desde frontend/localStorage/mock data hacia MySQL si aplica.
7. Preparar documentos legales y consentimientos.
8. Comprar dominio y VPS.
9. Configurar servidor.
10. Subir backend, frontend y base de datos.
11. Activar HTTPS y backups.
12. Probar con datos de prueba.
13. Validar con Rubi.
14. Publicar para uso real.

## 14. Frase simple para explicar el plan

> El VPS sera como la computadora en la nube donde vivira todo el sistema: la pagina visual, el backend y la base de datos. Antes de comprarlo debemos confirmar que el proyecto no tenga riesgos de licencia, que los documentos legales esten preparados y que sepamos como proteger los datos personales. Despues de comprarlo se configura Java, MySQL, Nginx, SSL, firewall y backups. Luego se sube Angular, Spring Boot y la base de datos exportada para que no se pierda informacion.
