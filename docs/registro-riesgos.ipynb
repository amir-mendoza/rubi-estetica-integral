````python
from IPython.display import Markdown, display

peso_probabilidad = {"Baja": 1, "Media": 2, "Alta": 3}
peso_impacto = {"Medio": 2, "Alto": 3, "Crítico": 4}
peso_estado = {"Bloquea producción": 0, "Pendiente": 1, "Mitigado": 2, "Resuelto": 3}

def puntaje(riesgo):
    return peso_probabilidad[riesgo["probabilidad"]] * peso_impacto[riesgo["impacto"]]

def mostrar_riesgos(estado=None):
    seleccion = riesgos if estado is None else [r for r in riesgos if r["estado"] == estado]
    seleccion = sorted(seleccion, key=lambda r: (peso_estado[r["estado"]], -puntaje(r), r["id"]))
    filas = ["| ID | Área | Riesgo | P x I | Estado | Acción | Responsable | Plazo | Evidencia |",
             "|---|---|---|---:|---|---|---|---|---|"]
    for r in seleccion:
        filas.append(f"| {r['id']} | {r['área']} | {r['riesgo']} | {puntaje(r)} | **{r['estado']}** | {r['acción']} | {r['responsable']} | {r['plazo']} | {r['evidencia']} |")
    display(Markdown("\n".join(filas)))

mostrar_riesgos()
````

````python
# Resumen automático. Cambia el estado en la matriz y vuelve a ejecutar.
from collections import Counter

resumen = Counter(r["estado"] for r in riesgos)
for estado in ["Bloquea producción", "Pendiente", "Mitigado", "Resuelto"]:
    print(f"{estado}: {resumen[estado]}")

print("\nBloqueos actuales:")
for r in riesgos:
    if r["estado"] == "Bloquea producción":
        print(f"- {r['id']} {r['área']}: {r['plazo']}")
````

## Decisiones técnicas aprobadas

| Capa | Decisión | Condición de seguridad |
|---|---|---|
| Frontend | Angular 22 | Sin secretos ni datos sensibles persistentes en el navegador |
| Backend | Spring Boot + Temurin 25 LTS | Spring Security, roles por endpoint/recurso, MFA administrativo y sesiones seguras |
| Datos | MySQL Community 8.4 LTS | Puerto 3306 cerrado; volumen persistente; mínimo privilegio; migraciones versionadas |
| Proxy | Nginx | Único servicio público, HTTPS y cabeceras de seguridad |
| Pagos | Izipay | Token, firma y webhook únicamente en backend; no almacenar tarjetas |
| Hosting | VPS Hostinger | Accesos individuales, MFA, firewall, parches, monitoreo y respaldo externo |

Arquitectura inicial:

```text
Internet -> HTTPS/Nginx -> Angular y /api -> Spring Boot -> MySQL privado
```

Angular, Spring Boot y MySQL pueden convivir inicialmente en un VPS. La base debe vivir fuera del directorio reemplazado durante el despliegue. Actualizar el frontend o el JAR nunca debe borrar MySQL.
